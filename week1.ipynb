## Step6: Checking and Handling Data Inconsistencies

```python
data['visit_date'] = pd.to_datetime(data['visit_date'], dayfirst=True)
```

```python
data['visit_date'].value_counts()
```

```python
data.dtypes
```

```python
device_mapping = {
    0: 'Desktop',
    1: 'Mobile',
    2: 'Tablet'
}
data['device_type'] = data['device_type'].map(device_mapping)
```

```python
data['device_type'].value_counts()
```

```python
users_mapping = {
    0: 'New',
    1: 'Returning'
}
data['user_type'] = data['user_type'].map(users_mapping)
```

```python
data['user_type'].value_counts()
```

```python
data['session_duration_bucket'].value_counts()
```

```python
data.dtypes
```

## Step7: Checking for Outliers
